# *Project Summary*

**End-to-End Retail Sales Analysis & Sales Prediction using Python, Machine Learning & Power BI** is a comprehensive Data Science project that analyzes retail sales data to uncover valuable business insights and predict future sales. The project follows the complete data science lifecycle, including data collection, data cleaning, preprocessing, exploratory data analysis (EDA), feature engineering, feature encoding, feature scaling, machine learning model development, model evaluation, and business intelligence reporting.

The **Superstore Retail Sales Dataset** was used to analyze customer purchasing behavior, product performance, regional sales trends, profitability, discounts, and shipping efficiency. Data preprocessing involved handling missing values, duplicate records, data type conversion, outlier detection using the IQR method, and creating new features such as **Shipping Days** and **Profit Margin**.

Model performance was evaluated using **MAE, MSE, RMSE, and R² Score**, and **Hyperparameter Tuning** was applied to optimize the best-performing model. Feature importance analysis was also performed to identify the variables that most significantly influence sales.

Finally, an interactive **Power BI Dashboard** was created to present KPIs, sales trends, customer insights, regional performance, and profitability through dynamic visualizations. The project concludes with business insights and strategic recommendations that support data-driven decision-making and improve retail business performance.

## **Exploratory Data Analysis & Visualization**

Exploratory Data Analysis (EDA) is the process of analyzing and visualizing a dataset to understand its structure, identify patterns, detect anomalies, and extract meaningful insights before building machine learning models. It helps in understanding the distribution of data, relationships between variables, trends, and business performance through statistical summaries and visualizations.

1. Monthly Sales Trend

```python
df1['Order Date'] = pd.to_datetime(df1['Order Date'], format='%d-%m-%Y')
df1['Month_Year'] = df1['Order Date'].dt.to_period('M')
monthly_sales = df1.groupby('Month_Year')['Sales'].sum().reset_index()
monthly_sales['Month_Year'] = monthly_sales['Month_Year'].astype(str)

plt.figure(figsize=(14,6))

plt.plot(
    monthly_sales['Month_Year'],
    monthly_sales['Sales'],
    marker='o',
    linewidth=2
)

plt.title("Monthly Sales Trend", fontsize=16)
plt.xlabel("Month-Year", fontsize=12)
plt.ylabel("Total Sales", fontsize=12)

plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

###  *Analysis*


* The monthly sales trend shows noticeable fluctuations over the observed period, indicating that sales vary significantly from month to month.

* During the early years (2011–2012), sales remained relatively moderate with a few sharp peaks, suggesting occasional periods of high demand.

* In 2013, sales began to improve, with several months recording higher revenue compared to previous years.

* A significant increase in sales is observed during late 2013, where monthly sales reached nearly ₹100,000, indicating strong business performance.

* Although there was a temporary decline in early 2014, sales recovered quickly and maintained an upward trend.

* The highest monthly sales were recorded in late 2014, exceeding ₹110,000, representing the best-performing month in the dataset.

* Overall, the trend indicates business growth over time, despite short-term fluctuations.

* The recurring peaks may suggest the presence of seasonal demand, promotional campaigns, or festive sales periods.

* The increasing sales toward the end of the timeline reflect improved customer demand and potentially effective business strategies.

Business Insights

* Sales show an overall upward growth trend from 2011 to 2014.
* 2014 recorded the highest sales compared to previous years.
* Monthly sales fluctuate significantly, indicating seasonal demand.
* Peak sales likely occurred due to festive seasons or promotional campaigns.
* Some months have low sales, highlighting opportunities for targeted marketing.
* Sales recover quickly after declines, showing strong business resilience.
* Higher sales in later years indicate increasing customer demand.
* Inventory should be increased before high-sales months.
* Focus marketing efforts on low-performing months to stabilize revenue.
* Historical sales trends can be used for accurate sales forecasting and business planning.

Label Encoding is a feature encoding technique used to convert categorical data into numerical values by assigning a unique integer to each category. It is commonly used when categorical variables have an inherent order or when machine learning algorithms require numerical input.

```python
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
```

```python
lr = LinearRegression()
knn = KNeighborsRegressor()
dt = DecisionTreeRegressor(random_state=42)
rf = RandomForestRegressor(random_state=42)
xgb = XGBRegressor(random_state=42)

lr.fit(X_train_scaled, y_train)
knn.fit(X_train_scaled, y_train)
dt.fit(X_train, y_train)
rf.fit(X_train, y_train)
xgb.fit(X_train, y_train)
```

```python
# MODEL EVALUATION
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

def evaluate(y_test, pred):
    print("MAE : ", mean_absolute_error(y_test, pred))
    print("MSE : ", mean_squared_error(y_test, pred))
    print("RMSE: ", np.sqrt(mean_squared_error(y_test, pred)))
    print("R2 Score: ", r2_score(y_test, pred))
```

## *Hyperparameter Tuning*

Hyperparameters are the settings or configurations of a machine learning model that are defined before training. Hyperparameter tuning involves testing different combinations of these settings to determine which combination produces the best model performance.**bold text**

```python
from xgboost import XGBRegressor
from sklearn.model_selection import RandomizedSearchCV

# Create model
xgb = XGBRegressor(
    objective="reg:squarederror",
    random_state=42
)

# Hyperparameter grid
param_grid = {
    "n_estimators": [100, 200, 300, 500],
    "learning_rate": [0.01, 0.05, 0.1, 0.2],
    "max_depth": [3, 5, 7, 9],
    "min_child_weight": [1, 3, 5],
    "subsample": [0.6, 0.8, 1.0],
    "colsample_bytree": [0.6, 0.8, 1.0],
    "gamma": [0, 0.1, 0.3, 0.5],
    "reg_alpha": [0, 0.01, 0.1, 1],
    "reg_lambda": [1, 2, 5]
}

# Random Search
random_search = RandomizedSearchCV(
    estimator=xgb,
    param_distributions=param_grid,
    n_iter=20,
    scoring="r2",
    cv=5,
    verbose=2,
    random_state=42,
    n_jobs=-1
)

# Fit
random_search.fit(X_train, y_train)

best_model = random_search.best_estimator_

# Best parameters
print("Best Parameters:")
print(random_search.best_params_)
```

```python
best_model = random_search.best_estimator_
```

```python
#Model prediction__After Hyperparameter Tuning
y_pred = best_model.predict(X_test)
```

```python
#  Model Evaluvation__After Hyperparameter Tuning
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

print("R2 Score :", r2_score(y_test, y_pred))
print("MAE      :", mean_absolute_error(y_test, y_pred))
print("MSE      :", mean_squared_error(y_test, y_pred))
print("RMSE     :", np.sqrt(mean_squared_error(y_test, y_pred)))
```
